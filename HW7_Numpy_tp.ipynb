**4. Індексація**

Створіть Numpy-масив `arr` зі значеннями [15, 22, 35, 44, 58, 67].

Виведіть (кожен пункт оцінюється в 0.5 бала):
- Перший елемент
- Останній елемент
- Елементи з індексами 1 до 3
- Елементи більші за 40

```python
arr = np.array([15, 22, 35, 44, 58, 67])
print(f"The first element of array arr is {arr[0]}")
print(f"The last element of array arr is {arr[-1]}")
print(f"The elements with indexes 1 to 3 of array arr are {arr[0:4]}")
mask = arr > 40
print(f"The elements > 30 of array arr are {arr[mask]}")
```
